# Prediction Notebook - SMS Spam Detector

This notebook loads the model we already trained (from `training.ipynb`) and uses it to
predict on new text. No training happens here — just loading and predicting.


## 1. Load the saved model, tokenizer, and config

```python
import pickle
import numpy as np
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.sequence import pad_sequences

model = load_model("model/model.h5")

with open("model/tokenizer.pkl", "rb") as f:
    tokenizer = pickle.load(f)

with open("model/config.pkl", "rb") as f:
    config = pickle.load(f)

max_len = config["max_len"]
print("Model, tokenizer, and config loaded!")
```

## 2. Make a simple prediction function

Same steps as training: tokenize the text -> pad it -> feed it to the model -> read the output.

```python
def predict_spam(text):
    seq = tokenizer.texts_to_sequences([text])
    padded = pad_sequences(seq, maxlen=max_len, padding='post', truncating='post')
    prob = model.predict(padded, verbose=0)[0][0]
    label = "SPAM" if prob > 0.5 else "HAM"
    return label, prob
```

## 3. Try it on new sentences

```python
test_messages = [
    "Congratulations, you've been selected for a free vacation!",
    "Hey, are we still meeting for lunch tomorrow?",
    "URGENT: your account will be suspended, click here now",
    "Can you send me the notes from class?",
    "You've won a $500 gift card, claim now!"
]

for msg in test_messages:
    label, prob = predict_spam(msg)
    print(f"{msg}\n -> {label} (confidence: {prob:.2%})\n")
```

## 4. Try your own message

Change the text below and re-run this cell to test any message you like.

```python
my_message = "Click here to claim your free iPhone now!"
label, prob = predict_spam(my_message)
print(f"{my_message}\n -> {label} (confidence: {prob:.2%})")
```
